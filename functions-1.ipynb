 ### Important Python built-in functions
# Here are some commonly used built-in functions:
# Function	Purpose	Example
# print()	Displays output	print("Hello")
# input()	Accepts user input	input("Enter name: ")
# type()	Checks data type	type(100)
# len()	Returns length	len("Python")
# int()	Converts to integer	int("25")
# float()	Converts to decimal	float("10.5")
# str()	Converts to string	str(100)
# sum()	Adds elements	sum([1,2,3])
# max()	Finds maximum	max(10,20,30)
# min()	Finds minimum	min(10,20,30)
# abs()	Returns absolute value	abs(-10)
# round()	Rounds a number	round(10.567, 2)
# sorted()	Sorts elements	sorted([3,1,2])
# range()	Generates a sequence	range(1,5)
# enumerate()	Gives index and value	enumerate(["a","b"])

### Type 1: No arguments and no return value

```python
def addition():
    a = 10
    b = 20
    print(a + b)

addition()
```

## Type 2: Arguments but no return value
You pass values to the function, but it only displays the result.



```python
def addition(a, b):
    print(a + b)

addition(10, 20)
addition(50, 30)
addition(100, 200)
```

```python
def addition(a, b):
    print(a + b)

addition(20, 20)
addition(50, 30)
addition(100, 200)

# a and b are parameters.
# 10 and 20 are arguments.
# print() displays the result.
```

## Type 3: No arguments but return value
### The function doesn't accept arguments but returns a value.

```python
def addition():
    a = 10
    b = 20
    return a + b

result = addition()
print(result)
```

### Type 4: Arguments and return value
